```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings  
warnings.filterwarnings('ignore')

df = pd.read_csv('/Users/mamatajoshi/Desktop/HCI/eda/tests/test_data/CovidData.csv', nrows=100000)

print(f'Dataset loaded: {df.shape[0]:,} rows, {df.shape[1]} columns')
print(df.head())
```

```python
# CLEANING STEP 8 — Remove impossible ages
print('--- Cleaning Step 8: Remove age outliers ---')
print(f'Rows before: {len(df):,}')

df['AGE'] = pd.to_numeric(df['AGE'], errors='coerce')
df = df[(df['AGE'] >= 0) & (df['AGE'] <= 120)]

print(f'Rows after: {len(df):,}')
print(f'Age range: {df["AGE"].min()} to {df["AGE"].max()}')
print('✅ Step 8 done!')
```

```python
# CLEANING STEP 9 — Replace invalid codes 97, 98, 99 with NaN
import numpy as np
print('--- Cleaning Step 9: Replace invalid codes ---')

cols = [c for c in df.columns if c != 'AGE']
total = 0
for col in cols:
    count = df[col].isin([97, 98, 99]).sum()
    total += count
    df[col] = df[col].replace([97, 98, 99], np.nan)

print(f'Total invalid values replaced with NaN: {total:,}')
print('✅ Step 9 done!')
```

```python
# CLEANING STEP 10 — Drop columns with more than 50% missing values
print('--- Cleaning Step 10: Drop columns >50% missing ---')
print(f'Columns before: {df.shape[1]}')

missing_pct = df.isnull().mean() * 100
cols_to_drop = missing_pct[missing_pct > 50].index.tolist()

if cols_to_drop:
    print(f'Dropping: {cols_to_drop}')
    df = df.drop(columns=cols_to_drop)
else:
    print('No columns exceeded 50% missing')

print(f'Columns after: {df.shape[1]}')
print(f'Final shape: {df.shape}')
print('✅ Step 10 done!')
print('🎉 All 3 cleaning steps complete!')
```
